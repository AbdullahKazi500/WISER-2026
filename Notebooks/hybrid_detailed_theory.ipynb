## Theoretical Preface — QAOA, Lagrangian Duals, and CVaR Risk Objectives

This notebook solves the same family of staffing QUBO as the PCE notebook, but with the other
major near-term quantum heuristic: the **Quantum Approximate Optimization Algorithm (QAOA)**,
augmented with three classical wrappers that are doing much of the practical work — a
**Lagrangian dual** outer loop, a **CVaR (Conditional Value at Risk)** inner objective, and a
**warm start** from a classical greedy solution. Understanding each layer is necessary to read
the code correctly.

### 1. QAOA in one paragraph
Given a cost Hamiltonian $H_C$ (the Ising form of the QUBO) and a mixer Hamiltonian
$H_B=\sum_i X_i$, QAOA prepares
$$
|\psi(\boldsymbol\gamma,\boldsymbol\beta)\rangle
= e^{-i\beta_p H_B}e^{-i\gamma_p H_C}\cdots e^{-i\beta_1 H_B}e^{-i\gamma_1 H_C}\,|+\rangle^{\otimes n},
$$
alternating $p$ layers of cost-Hamiltonian phase kicks and mixer-Hamiltonian rotations, and
classically optimizes $(\boldsymbol\gamma,\boldsymbol\beta)$ to minimize
$\langle\psi|H_C|\psi\rangle$ (or, here, a **CVaR** statistic of sampled energies — see below).
Measuring the optimized circuit in the computational basis yields bitstrings that are biased
toward low-$H_C$-energy (i.e. low QUBO-cost) solutions.

### 2. Why CVaR instead of the mean energy
The plain QAOA objective $\langle H_C\rangle$ is an *average* over all $2^n$ measurement
outcomes, including many bad ones — for small $p$ this average is a poor proxy for "can we find
a good bitstring." **CVaR$_\alpha$** instead only averages the best $\alpha$-fraction of the
sampled energy distribution:
$$
\mathrm{CVaR}_\alpha(E) = \frac{1}{\alpha}\int_0^{\alpha} F_E^{-1}(u)\,du,
$$
i.e. the expected value of the energy *conditional on being in the best $\alpha$ quantile*. This
notebook uses $\alpha=0.25$: the loss function only "cares" about the best quarter of sampled
bitstrings, which is much more aligned with the actual goal (find *a* good staffing plan) than
optimizing the full-distribution mean.

### 3. Warm-starting QAOA
Instead of starting every qubit in the uniform superposition $|+\rangle$ (mixer ground state,
$R_Y(\pi/2)$), each qubit $i$ is initialized at $R_Y(\theta_i^{\mathrm{ws}})$ where $\theta_i^{\mathrm{ws}}$
encodes a classical **greedy solution bit** as a *biased* superposition
($2\arcsin\sqrt{v}$ with $v$ close to 0 or 1, clipped by $\epsilon$ to avoid a deterministic
start). The mixer is correspondingly modified to rotate around each qubit's *own* biased axis
rather than the uniform $X$ axis — implemented here as conjugating the $R_Z(-2\beta)$ mixer phase
by $R_Y(\mp\theta_i^{\mathrm{ws}})$. This is the **warm-start QAOA** construction of Egger et
al.: it biases the search to explore *near* a known-good classical solution rather than from
scratch.

### 4. The outer Lagrangian dual
Hard coverage constraints $\mathrm{cov}_t\ge R_t^{\mathrm{SLA}}$ are softened into the QUBO with
*fixed* penalty weights $P_{\mathrm{SLA}}, P_{\mathrm{ASA}}$ — but fixed weights can
over/under-penalize different intervals. `dual_averaging_loop` instead runs a **subgradient
ascent on Lagrange multipliers** $\lambda_t$ (one per interval), increasing $\lambda_t$ for
intervals whose coverage keeps falling short across outer iterations, and passing the
reweighted QUBO back into (warm-started, CVaR-)QAOA. This is a textbook **primal–dual** hybrid
loop: the quantum step solves a primal-like relaxation; the classical step adjusts dual
multipliers based on constraint violation.

## 1. Mathematical formulation

### Decision variables (binary)
Unary encoding of agents on each shift:

$$
x_{s,k}\in\{0,1\},\qquad
n_s=\sum_{k=0}^{N_{\max}-1}x_{s,k}\in\{0,\ldots,N_{\max}\}
$$

Coverage of interval \(t\):

$$
\mathrm{cov}_t=\sum_{s}C_{s,t}\,n_s
$$

where \(C_{s,t}\in\{0,1\}\) is the shift–interval incidence (0 during breaks).

### Linear constraints
$$
\begin{align}
\mathrm{cov}_t &\ge R_t^{\mathrm{SLA}} && \text{(Erlang agents for SLA ≥ 80\%)}\\
\mathrm{cov}_t &\ge R_t^{\mathrm{ASA}} && \text{(Erlang agents for ASA ≤ 25 s)}\\
n_s &\le N_{\max}\\
\text{OT} &\le \text{OT}_{\max}
\end{align}
$$

In the quantum step these are **softened** (quadratic shortfall penalties).  
They are **hard-checked** after decode via true Erlang simulation.

### Quadratic objective (minimize)
$$
h(x)=\sum_s c_s n_s
+P_{\mathrm{SLA}}\sum_t\bigl(\mathrm{cov}_t-R_t^{\mathrm{SLA}}\bigr)_-^{2}
+P_{\mathrm{ASA}}\sum_t w_t\bigl(\mathrm{cov}_t-R_t^{\mathrm{ASA}}\bigr)_-^{2}
$$

Implemented in QUBO form without slack variables (Cutting-Slack style).


### From continuous constraint to discrete penalty — same derivation as the PCE notebook

Exactly as in the PCE staffing notebook, $(\mathrm{cov}_t-R_t)^2$ is expanded in binary
variables $x_i\in\{0,1\}$ (using $x_i^2=x_i$) into linear + quadratic QUBO coefficients; here the
weight on interval $t$'s SLA term additionally carries the **dual multiplier**
$\lambda_t$ (fixed at $\lambda_t\equiv1$ inside a single QAOA call, but updated between calls by
the outer Lagrangian loop in §7):
$$
h_\lambda(x)=\sum_s c_s n_s
+P_{\mathrm{SLA}}\sum_t\lambda_t(\mathrm{cov}_t-R_t^{\mathrm{SLA}})^2
+P_{\mathrm{ASA}}\sum_t w_t(\mathrm{cov}_t-R_t^{\mathrm{ASA}})^2 .
$$
No slack qubits are introduced for the inequality constraints — the one-sided "shortfall" is
handled the same signed-residual way as in the PCE notebook (never over-penalizing
*over*-staffing, only *under*-staffing, is an approximation the classical repair step corrects
for exactly).

## 2. Erlang-C service metrics


### Erlang-C, plus two derived operational metrics unique to this notebook

`service_level`, `asa_seconds` are identical in form to the PCE notebook (see its §2 theory
block for the full Erlang-C derivation). This notebook additionally derives:

**Abandonment rate** (heuristic linear-in-delay model, capped at 50%):
$$
\mathrm{Ab}(n,\lambda) = \min\!\Bigl(0.5,\ \beta_0 + 0.01\cdot\frac{\max(0,\ \mathrm{ASA}(n,\lambda)-10)}{10}\Bigr),
$$
i.e. abandonment grows linearly once ASA exceeds 10s past the baseline abandonment rate
$\beta_0=$ `ABANDON_RATE_BASE`.

**Occupancy** (fraction of paid agent-time spent actively handling calls, capped at 100%):
$$
\mathrm{Occ}(n,\lambda) = \min\!\left(1,\ \frac{A}{n}\right), \qquad A=\frac{\lambda}{\Delta t}h.
$$
Occupancy near 1 signals agents are saturated (little slack for load spikes); occupancy far
below 1 signals over-staffing relative to the offered load $A$ — the trade-off any cost-optimal
staffing plan is implicitly navigating.

```python

def erlang_c(n, A):
    if n <= 0 or n <= A: return 1.0
    try:
        rho = A / n; B = 1.0
        for k in range(1, n + 1): B = 1.0 + (k / A) * B
        return 1.0 / (1.0 + (1.0 - rho) * (B - 1.0) / rho) if rho > 0 else 0.0
    except Exception:
        return 1.0

def service_level(n, arrivals, tau=20.0):
    if arrivals <= 0: return 1.0
    if n <= 0: return 0.0
    A = (arrivals / INTERVAL_SEC) * AHT
    if n <= A: return 0.0
    pw = erlang_c(n, A)
    return float(max(0.0, min(1.0, 1.0 - pw * exp(-(n - A) * tau / AHT))))

def asa_seconds(n, arrivals):
    if arrivals <= 0: return 0.0
    if n <= 0: return 999.0
    A = (arrivals / INTERVAL_SEC) * AHT
    if n <= A: return 999.0
    return float(erlang_c(n, A) * AHT / (n - A))

def abandonment(n, arrivals):
    asa = asa_seconds(n, arrivals)
    return float(min(0.5, ABANDON_RATE_BASE + 0.01 * max(0, asa - 10) / 10))

def occupancy(n, arrivals):
    if n <= 0: return 1.0
    A = (arrivals / INTERVAL_SEC) * AHT
    return float(min(1.0, A / n))

def required_agents(arrivals, target_sl=None, target_asa=None, max_n=50):
    if arrivals <= 0: return 0
    n = max(1, int(ceil(arrivals * AHT / INTERVAL_SEC)))
    while n <= max_n:
        ok1 = True if target_sl is None else service_level(n, arrivals) >= target_sl - 1e-6
        ok2 = True if target_asa is None else asa_seconds(n, arrivals) <= target_asa + 1e-6
        if ok1 and ok2: return n
        n += 1
    return max_n

print("Erlang helpers ready")
```

Below: all four Erlang-family response curves (Erlang-C, SL, ASA, abandonment) plus occupancy,
plotted together for a representative offered load, to visualize *why* there is a single
"knee" region where adding one more agent produces a large SLA/ASA/abandonment improvement, and
diminishing returns beyond it — the qualitative shape every constraint-generation step in this
notebook is built on.

```python
# --- Scientific plot: full Erlang-family operational curves vs staffing level ---
lam_rep = 20.0  # representative interval arrival rate for illustration
A_rep = (lam_rep / INTERVAL_SEC) * AHT
n_grid = np.arange(int(np.ceil(A_rep)) + 1, int(np.ceil(A_rep)) + 20)

sl_c  = [service_level(n, lam_rep) for n in n_grid]
asa_c = [asa_seconds(n, lam_rep) for n in n_grid]
ab_c  = [abandonment(n, lam_rep) for n in n_grid]
oc_c  = [occupancy(n, lam_rep) for n in n_grid]

fig, ax1 = plt.subplots(figsize=(9, 4.5))
ax1.plot(n_grid, sl_c, "o-", color="#4C72B0", label="Service level")
ax1.plot(n_grid, oc_c, "^-", color="#55A868", label="Occupancy")
ax1.axhline(TARGET_SL, color="red", ls="--", lw=1)
ax1.set_xlabel("agents $n$"); ax1.set_ylabel("fraction")
ax1.yaxis.set_major_formatter(mticker.PercentFormatter(1.0))

ax2 = ax1.twinx()
ax2.plot(n_grid, asa_c, "s-", color="#DD8452", label="ASA (s)")
ax2.plot(n_grid, np.array(ab_c) * 100, "d-", color="#8172B2", label="Abandon (%, x1)")
ax2.axhline(TARGET_ASA, color="red", ls=":", lw=1)
ax2.set_ylabel("ASA (s) / Abandonment (%)"); ax2.set_yscale("log")

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, fontsize=8, loc="center right")
ax1.set_title(rf"Erlang-family operating curves at $A={A_rep:.2f}$ Erlangs")
plt.tight_layout(); plt.show()
```

## 3. Synthetic historical arrivals (TOD × queue × channel)


### Synthetic arrival-process theory

`generate_history` simulates $\lambda_{d,t,q,c}$ (arrivals for day $d$, interval $t$, queue $q$,
channel $c$) as **Poisson-distributed counts** whose intensity is a *multiplicative* model
$$
\lambda_{d,t,q,c} = \mathrm{Poisson}\!\Bigl(\Lambda_0\cdot\underbrace{f_{\mathrm{tod}}(t)}_{\text{time-of-day}}
\cdot\underbrace{f_{\mathrm{dow}}(d)\cdot u_d}_{\text{day-of-week}\times\text{noise}}
\cdot\underbrace{f_{\mathrm{ch}}(c)}_{\text{channel mix}}
\cdot\underbrace{f_{q}(q)}_{\text{queue mix}}\Bigr),
$$
with a base rate $\Lambda_0=18$, a bell-shaped intraday profile `base_tod` peaking mid-morning,
a day-of-week multiplier `dow` (weekday-heavy, weekend-light), i.i.d. daily multiplicative noise
$u_d\sim\mathrm{Unif}(0.92,1.08)$, and fixed channel/queue scale factors. This is a standard way
to generate **realistic, seasonally-structured Poisson arrival data** for queueing simulation
without needing real historical call logs.

```python

def generate_history(n_days=14, n_intervals=12, n_queues=2, n_channels=2, seed=7):
    rng = np.random.default_rng(seed)
    base_tod = np.array([0.35, 0.55, 0.80, 1.05, 1.20, 1.10, 0.95, 0.90, 1.05, 0.95, 0.65, 0.40])
    if n_intervals != 12:
        base_tod = np.interp(np.linspace(0, 11, n_intervals), np.arange(12), base_tod)
    dow = np.array([1.0, 1.05, 1.0, 0.98, 0.95, 0.70, 0.55])
    queues = [f"Q{q}" for q in range(n_queues)]
    channels = ["voice", "chat"][:n_channels]
    rows = []
    for day in range(n_days):
        dscale = dow[day % 7] * rng.uniform(0.92, 1.08)
        for t in range(n_intervals):
            for qi, q in enumerate(queues):
                for ch in channels:
                    ch_scale = 1.0 if ch == "voice" else 0.45
                    q_scale = 1.0 if qi == 0 else 0.7
                    lam = 18.0 * base_tod[t] * dscale * ch_scale * q_scale
                    rows.append(dict(day=day, interval=t, queue=q, channel=ch,
                                     arrivals=int(rng.poisson(max(1.0, lam)))))
    return pd.DataFrame(rows)

hist = generate_history()
print(f"History: {len(hist)} rows | total arrivals={hist['arrivals'].sum()}")
display(hist.head(8))
```

## 4. Forecast, shifts, breaks, OT — instance builders


### Shift-cover incidence and cost-model theory

`build_shift_cover` assigns each shift $s$ a contiguous span of intervals
$[a_s,b_s)\subset\{0,\ldots,n_{\mathrm{int}}-1\}$ (with one break interval zeroed out at the
midpoint, mirroring a mandated meal/rest break), producing the same kind of binary incidence
matrix $C_{s,t}$ used throughout both notebooks. `shift_costs` sets
$c_s = c_0(1+0.05s)$ — later shifts cost marginally more (evening/night differential) — and
optionally multiplies the **last** shift's cost by `OT_MULT` to represent an overtime-eligible
shift. `make_instance` assembles a **small** (12 intervals, 3 shifts, $N_{\max}=4$) and a
**large** (24 intervals, 4 shifts, $N_{\max}=3$) test instance, both re-using the Erlang-based
`required_agents` to derive $R_t^{\mathrm{SLA}}, R_t^{\mathrm{ASA}}$ — i.e. both instances share
the exact same constraint-generation pipeline validated above, just at different scale.

```python

def build_shift_cover(n_intervals, n_shifts=3, break_frac=0.1):
    C = np.zeros((n_shifts, n_intervals), dtype=float)
    spans = [
        (0, int(0.65 * n_intervals)),
        (int(0.25 * n_intervals), int(0.85 * n_intervals)),
        (int(0.50 * n_intervals), n_intervals),
        (int(0.10 * n_intervals), int(0.75 * n_intervals)),
    ]
    for s, (a, b) in enumerate(spans[:n_shifts]):
        C[s, a:b] = 1.0
        mid = (a + b) // 2
        if break_frac > 0 and 0 <= mid < n_intervals:
            C[s, mid] = 0.0
    return C

def shift_costs(n_shifts, base=COST_BASE, ot_shift_idx=None):
    c = np.array([base * (1.0 + 0.05 * s) for s in range(n_shifts)])
    if ot_shift_idx is not None:
        for s in ot_shift_idx:
            c[s] *= OT_MULT
    return c

def make_instance(size="small", seed=7):
    rng = np.random.default_rng(seed)
    if size == "small":
        n_int, n_shifts, nmax = 12, 3, 4
        base = np.array([5, 9, 14, 18, 22, 20, 16, 15, 18, 16, 11, 6], dtype=float)
        demand = np.maximum(1, np.round(base * rng.uniform(0.92, 1.08, n_int))).astype(int)
        hist_s = generate_history(14, n_int, 2, 2, seed)
        br = 0.08
    else:
        n_int, n_shifts, nmax = 24, 4, 3
        base12 = np.array([5, 9, 14, 18, 22, 20, 16, 15, 18, 16, 11, 6], dtype=float)
        base = np.interp(np.linspace(0, 11, n_int), np.arange(12), base12)
        demand = np.maximum(1, np.round(base * rng.uniform(0.90, 1.05, n_int) * 0.70)).astype(int)
        hist_s = generate_history(21, n_int, 2, 2, seed)
        br = 0.0
    C = build_shift_cover(n_int, n_shifts, break_frac=br)
    costs = shift_costs(n_shifts, ot_shift_idx=[n_shifts - 1] if n_shifts > 2 else None)
    R_sla = np.array([required_agents(a, target_sl=TARGET_SL) for a in demand])
    R_asa = np.array([required_agents(a, target_asa=TARGET_ASA) for a in demand])
    return dict(size=size, n_int=n_int, n_shifts=n_shifts, nmax=nmax,
                demand=demand, C=C, costs=costs, R_sla=R_sla, R_asa=R_asa,
                hist=hist_s, qubits=n_shifts * nmax)

inst_s = make_instance("small")
inst_l = make_instance("large")
print(f"SMALL: intervals={inst_s['n_int']} shifts={inst_s['n_shifts']} Nmax={inst_s['nmax']} "
      f"qubits={inst_s['qubits']} demand_sum={inst_s['demand'].sum()}")
print(f"  R_SLA={inst_s['R_sla'].tolist()}")
print(f"LARGE: intervals={inst_l['n_int']} shifts={inst_l['n_shifts']} Nmax={inst_l['nmax']} "
      f"qubits={inst_l['qubits']} demand_sum={inst_l['demand'].sum()}")
```

## 5. Evaluation, simulation, classical greedy


### The classical greedy baseline as a matroid-flavored covering heuristic

`greedy_warmstart` is a **greedy set-cover-style** heuristic: at each step it adds one agent to
the shift $s$ maximizing "remaining coverage bought per dollar,"
$\mathrm{score}(s)=\dfrac{C_s\cdot(\mathrm{rem}>0)}{c_s}$, where $\mathrm{rem}_t=\max(0,
R_t-\mathrm{cov}_t)$ is the still-unmet requirement. This is the textbook greedy approximation
algorithm for weighted set cover (known to be within a $\ln(\max_t R_t)$ factor of optimal for
*pure* covering problems, though the staffing problem's joint SLA+ASA+cost structure is not
exactly weighted set cover, so this bound is only a heuristic motivation, not a guarantee here).
`repair_n` / the ASA-push loop then locally patches any residual shortfall by the same
greedy-per-dollar logic, and this combination *is* the classical baseline every quantum-hybrid
result in this notebook is benchmarked against.

```python

def coverage(n, C):
    return C.T @ np.asarray(n, dtype=float)

def evaluate_plan(n, inst):
    n = np.asarray(n, dtype=int)
    cov = coverage(n, inst["C"])
    demand = inst["demand"]
    cost = float(np.dot(inst["costs"], n))
    sls, asas, abnds, occs, calls = [], [], [], [], []
    for t, arr in enumerate(demand):
        a = int(round(cov[t]))
        sls.append(service_level(a, int(arr)))
        asas.append(asa_seconds(a, int(arr)))
        abnds.append(abandonment(a, int(arr)))
        occs.append(occupancy(a, int(arr)))
        calls.append(int(arr))
    w = np.maximum(np.array(calls, float), 1e-6)
    sla = float(np.average(sls, weights=w))
    asa = float(np.average(asas, weights=w))
    return dict(
        n=n.tolist(), cost=cost, sla=sla, asa=asa,
        abandon=float(np.average(abnds, weights=w)),
        util=float(np.mean(occs)), agents=int(n.sum()),
        meets_sla=sla >= TARGET_SL - 1e-3,
        meets_asa=asa <= TARGET_ASA + 1e-3,
        meets_both=(sla >= TARGET_SL - 1e-3) and (asa <= TARGET_ASA + 1e-3),
        coverage=cov.tolist(), interval_sla=sls, interval_asa=asas,
    )

def greedy_warmstart(inst):
    n_shifts, nmax = inst["n_shifts"], inst["nmax"]
    C, costs = inst["C"], inst["costs"]
    R = np.maximum(inst["R_sla"], inst["R_asa"]).astype(float)
    n = np.zeros(n_shifts, dtype=int)
    rem = R.copy()
    for _ in range(int(rem.sum()) + n_shifts + 5):
        if rem.max() <= 0: break
        best_s, best_sc = None, -1e99
        for s in range(n_shifts):
            if n[s] >= nmax: continue
            sc = (C[s] @ (rem > 0).astype(float)) / (costs[s] + 1e-6)
            if sc > best_sc: best_sc, best_s = sc, s
        if best_s is None or best_sc <= 0: break
        n[best_s] += 1
        rem = np.maximum(0, rem - C[best_s])
    return n

def repair_n(n, inst):
    n = np.asarray(n, dtype=int).copy()
    C, nmax = inst["C"], inst["nmax"]
    R = inst["R_sla"].astype(float)
    cov = coverage(n, C)
    for _ in range(int(np.maximum(0, R - cov).sum()) + 8):
        short = R - cov
        if short.max() <= 0: break
        t = int(np.argmax(short))
        best_s = next((s for s in range(inst["n_shifts"]) if n[s] < nmax and C[s, t] > 0.5), None)
        if best_s is None: break
        n[best_s] += 1
        cov = coverage(n, C)
    demand = inst["demand"]
    for _ in range(20):
        asas = [asa_seconds(int(round(cov[t])), int(demand[t])) for t in range(len(demand))]
        if max(asas) <= TARGET_ASA: break
        t = int(np.argmax(asas))
        best_s = next((s for s in range(inst["n_shifts"]) if n[s] < nmax and C[s, t] > 0.5), None)
        if best_s is None: break
        n[best_s] += 1
        cov = coverage(n, C)
    return n

def classical_greedy(inst):
    return evaluate_plan(repair_n(greedy_warmstart(inst), inst), inst)

g_s = classical_greedy(inst_s)
g_l = classical_greedy(inst_l)
print(f"Greedy SMALL: cost=${g_s['cost']:.0f} SLA={g_s['sla']:.1%} ASA={g_s['asa']:.1f}s both={g_s['meets_both']}")
print(f"Greedy LARGE: cost=${g_l['cost']:.0f} SLA={g_l['sla']:.1%} ASA={g_l['asa']:.1f}s both={g_l['meets_both']}")
```

## 6. Slack-free dual QUBO + CVaR-QAOA

### QUBO (no slack qubits)
$$
h(x)=\sum_s c_s n_s
+P_{\mathrm{SLA}}\sum_t\lambda_t(\mathrm{cov}_t-R_t^{\mathrm{SLA}})^2
+P_{\mathrm{ASA}}\sum_t w_t(\mathrm{cov}_t-R_t^{\mathrm{ASA}})^2
$$

### CVaR objective
$$
\mathrm{CVaR}_\alpha=\frac1\alpha\int_0^\alpha F_E^{-1}(u)\,du
$$
focuses the variational loop on the low-energy tail of samples.


### Full QAOA circuit construction, term by term (matching `run_cvar_qaoa` exactly)

Let $\{h_i\}, \{J_{ij}\}$ be the Ising coefficients from `qubo_to_ising` (derived from the QUBO
exactly as in the PCE notebook's linear-algebra derivation, then converted via
$x_i=\tfrac{1-z_i}{2}$). One QAOA layer applies:

1. **Cost unitary** $e^{-i\gamma H_C}$, realized gate-by-gate as
   $R_Z(2\gamma h_i)$ on each qubit $i$ (`qc.rz(2*gammas[layer]*coef, i)`) and
   $R_{ZZ}(2\gamma J_{ij})$ on each coupled pair (`qc.rzz(...)`) — the standard Trotterized
   decomposition of a diagonal Ising Hamiltonian into native two-qubit $ZZ$ rotations.
2. **Warm-start mixer** $e^{-i\beta H_B^{\mathrm{ws}}}$, realized as
   $R_Y(-\theta_i^{\mathrm{ws}})\to R_Z(-2\beta)\to R_Y(\theta_i^{\mathrm{ws}})$ — this
   conjugation rotates the standard $Z$-mixer axis to align with each qubit's warm-start bias
   axis $\theta_i^{\mathrm{ws}}=2\arcsin\sqrt{v_i}$ (computed once, before the layer loop, from
   the greedy warm-start bits).

Rather than treating $(\gamma_1,\ldots,\gamma_p,\beta_1,\ldots,\beta_p)$ as $2p$ independent free
parameters, `run_cvar_qaoa` uses a common **linear-ramp schedule** parameterized by only 2 scalars
$(\Delta\gamma,\Delta\beta)$:
$$
\gamma_\ell = \ell\cdot\frac{\Delta\gamma}{p}\ (\ell=1,\ldots,p), \qquad
\beta_\ell = (p-\ell+1)\cdot\frac{\Delta\beta}{p},
$$
a common trick (linear/annealing-inspired schedules) that collapses the optimization search
space from $2p$ to 2 dimensions — tractable for a handful of COBYLA iterations even though the
circuit itself may have several layers $p$.

### The CVaR loss in code
`cvar_obj(params)` samples the bound circuit (60 shots, cheap for the *inner* optimization
loop), computes each sampled bitstring's exact Ising energy via `energy_bits`, and calls
`compute_cvar(probs, values, alpha)` — a direct implementation of the $\mathrm{CVaR}_\alpha$
integral above using the empirical (sampled) distribution: sort by energy, accumulate
probability mass up to $\alpha$, average the included energies weighted by their (possibly
truncated) probability.

```python

def build_dual_qubo(inst, P_sla=400.0, P_asa=120.0, lam=None):
    n_shifts, nmax = inst["n_shifts"], inst["nmax"]
    C, costs = inst["C"], inst["costs"]
    R_sla, R_asa = inst["R_sla"], inst["R_asa"]
    n_int = inst["n_int"]
    if lam is None: lam = np.ones(n_int)
    idx, n = {}, 0
    for s in range(n_shifts):
        for k in range(nmax):
            idx[("x", s, k)] = n; n += 1
    n_vars = n
    Q = defaultdict(float)
    def add(i, j, c):
        if i > j: i, j = j, i
        Q[(i, j)] += c
    for s in range(n_shifts):
        for k in range(nmax):
            add(idx[("x", s, k)], idx[("x", s, k)], float(costs[s]))
    for t in range(n_int):
        terms = [idx[("x", s, k)] for s in range(n_shifts) if C[s, t] > 0.5 for k in range(nmax)]
        R, w = float(R_sla[t]), float(lam[t]) * P_sla
        for i1 in terms: add(i1, i1, w * (1 - 2 * R))
        for a in range(len(terms)):
            for b in range(a + 1, len(terms)):
                add(terms[a], terms[b], w * 2)
    for t in range(n_int):
        terms = [idx[("x", s, k)] for s in range(n_shifts) if C[s, t] > 0.5 for k in range(nmax)]
        R = float(R_asa[t])
        w = P_asa * (1.0 + 0.4 * (R / max(1.0, float(R_asa.max()))))
        for i1 in terms: add(i1, i1, w * (1 - 2 * R))
        for a in range(len(terms)):
            for b in range(a + 1, len(terms)):
                add(terms[a], terms[b], w * 2)
    return dict(Q), idx, n_vars

def qubo_to_ising(Q, n_vars):
    h = np.zeros(n_vars); J = {}; offset = 0.0
    for (i, j), q in Q.items():
        if i == j:
            offset += q / 2; h[i] -= q / 2
        else:
            offset += q / 4; h[i] -= q / 4; h[j] -= q / 4
            J[(i, j)] = J.get((i, j), 0.0) + q / 4
    ising = {(i, i): h[i] for i in range(n_vars) if abs(h[i]) > 1e-12}
    ising.update({(i, j): c for (i, j), c in J.items() if abs(c) > 1e-12})
    return ising, offset

def bits_to_n(bits, idx, n_shifts, nmax):
    n = np.zeros(n_shifts, dtype=int)
    for s in range(n_shifts):
        n[s] = sum(bits[idx[("x", s, k)]] for k in range(nmax))
    return n

def energy_bits(bits, ising, offset):
    zs = [1 - 2 * int(b) for b in bits]
    v = offset
    for (i, j), c in ising.items():
        v += c * zs[i] if i == j else c * zs[i] * zs[j]
    return float(v)

def compute_cvar(probs, values, alpha):
    order = np.argsort(values)
    probs = np.asarray(probs)[order]; vals = np.asarray(values)[order]
    cvar, total = 0.0, 0.0
    for p, v in zip(probs, vals):
        if total + p > alpha: p = alpha - total
        total += p; cvar += p * v
        if total >= alpha - 1e-12: break
    return cvar / alpha

Qs, idxs, nvs = build_dual_qubo(inst_s)
print(f"SMALL QUBO: {nvs} qubits (slack=0) | Ising terms={len(qubo_to_ising(Qs, nvs)[0])}")
```

```python

from qiskit import QuantumCircuit, transpile
from qiskit.circuit import ParameterVector
from qiskit_aer import AerSimulator

def run_cvar_qaoa(inst, lam=None, alpha=0.25, p=1, shots=350, maxiter=3):
    Q, idx, n_vars = build_dual_qubo(inst, lam=lam)
    ising, offset = qubo_to_ising(Q, n_vars)
    n_shifts, nmax = inst["n_shifts"], inst["nmax"]
    n0 = greedy_warmstart(inst)
    eps = 0.12
    ws = []
    for s in range(n_shifts):
        for k in range(nmax):
            bit = 1 if k < n0[s] else 0
            v = min(max(float(bit), eps), 1 - eps)
            ws.append(2 * np.arcsin(np.sqrt(v)))
    gammas = ParameterVector("g", p)
    betas = ParameterVector("b", p)
    qc = QuantumCircuit(n_vars)
    for i, th in enumerate(ws): qc.ry(th, i)
    for layer in range(p):
        for (i, j), coef in ising.items():
            if i == j: qc.rz(2 * gammas[layer] * float(coef), i)
            else: qc.rzz(2 * gammas[layer] * float(coef), i, j)
        for i, th in enumerate(ws): qc.ry(-th, i)
        qc.rz(-2 * betas[layer], range(n_vars))
        for i, th in enumerate(ws): qc.ry(th, i)
    qc.measure_all()
    backend = AerSimulator(method="matrix_product_state")

    def sample(params, sh):
        dg, db = params[0], params[1] if len(params) > 1 else params[0]
        gv = np.arange(1, p + 1) * dg / p
        bv = np.arange(1, p + 1)[::-1] * db / p
        bind = {gammas[i]: gv[i] for i in range(p)}
        bind.update({betas[i]: bv[i] for i in range(p)})
        bound = qc.assign_parameters(bind)
        try: tqc = transpile(bound, backend=backend, optimization_level=0)
        except Exception: tqc = bound
        return backend.run(tqc, shots=sh).result().get_counts()

    def cvar_obj(params):
        counts = sample(params, 60)
        tot = sum(counts.values())
        probs, vals = [], []
        for bs, cnt in counts.items():
            bits = [int(b) for b in bs[::-1].zfill(n_vars)]
            probs.append(cnt / tot)
            vals.append(energy_bits(bits, ising, offset))
        return compute_cvar(probs, vals, alpha)

    t0 = time.time()
    res = minimize(cvar_obj, x0=np.array([0.55, 0.40]),
                   bounds=[(0.05, 1.3), (0.05, 1.0)],
                   method="COBYLA", options={"maxiter": maxiter})
    counts = sample(res.x, shots)
    qtime = time.time() - t0
    ranked = sorted(counts.items(), key=lambda kv: -kv[1])[:60]
    cands = []
    for bs, cnt in ranked:
        bits = [int(b) for b in bs[::-1].zfill(n_vars)]
        n = repair_n(bits_to_n(bits, idx, n_shifts, nmax), inst)
        m = evaluate_plan(n, inst)
        m["shots"] = cnt
        cands.append(m)
    cands.sort(key=lambda m: (not m["meets_both"], not m["meets_sla"], m["cost"], m["asa"]))
    both = [c for c in cands if c["meets_both"]]
    if both:
        best, tag = both[0], "BOTH"
    else:
        sla_ok = [c for c in cands if c["meets_sla"]]
        best, tag = (min(sla_ok, key=lambda m: (m["cost"], m["asa"])), "SLA_ONLY") if sla_ok else (cands[0], "NONE")
    best.update(dict(time=qtime, tag=tag, alpha=alpha, n_both=len(both), qubits=n_vars))
    return best, cands

print("CVaR-QAOA solver ready")
```

## 7. Outer Lagrangian dual + full hybrid on small & large


### Subgradient (dual-averaging) update, matched to the code

For the Lagrangian relaxation of $\mathrm{cov}_t\ge R_t^{\mathrm{SLA}}$ with multiplier
$\lambda_t\ge0$, the standard subgradient ascent step on the *dual* function
$g(\lambda)=\min_x\bigl[h_\lambda(x)\bigr]$ is
$$
\lambda_t \leftarrow \max\bigl(0,\ \lambda_t + \eta\cdot(R_t^{\mathrm{SLA}} - \mathrm{cov}_t(x^\star(\lambda)))\bigr),
$$
i.e. increase $\lambda_t$ when the current QAOA-found solution under-covers interval $t$
(subgradient of $\lambda_t(\mathrm{cov}_t-R_t)^2$-type penalty w.r.t. $\lambda_t$ evaluated at
the constraint violation). `dual_averaging_loop` implements a **Cesàro-averaged** variant:
it accumulates $\text{cum\_sub}=\sum_{k'\le k}(R_t-\mathrm{cov}_t^{(k')})$ across outer
iterations $k=1,\ldots,\texttt{max\_outer}$ and sets
$$
\lambda_t^{(k)} = \max\Bigl(0.1,\ 1 + \eta\cdot\frac{\text{cum\_sub}_t}{k}\Bigr), \qquad \eta=\texttt{alpha\_step},
$$
i.e. an average of past subgradients rather than the instantaneous one — a standard
variance-reduction trick (**dual averaging**, Nesterov) for noisy subgradients, which is exactly
the situation here since each "subgradient" comes from a *sampled*, warm-started QAOA solve
rather than an exact minimization.

```python

def dual_averaging_loop(inst, max_outer=3, alpha_step=0.12):
    n_int = inst["n_int"]
    lam = np.ones(n_int)
    best_plan = None
    cum_sub = np.zeros(n_int)
    t_total = 0.0
    history = []
    for k in range(1, max_outer + 1):
        plan, _ = run_cvar_qaoa(inst, lam=lam, alpha=0.25, shots=300, maxiter=2)
        t_total += plan["time"]
        cov = np.array(plan["coverage"])
        sub = inst["R_sla"] - cov
        cum_sub += sub
        lam = np.maximum(0.1, 1.0 + alpha_step * cum_sub / k)
        history.append(dict(iter=k, cost=plan["cost"], sla=plan["sla"], asa=plan["asa"], both=plan["meets_both"]))
        if plan["meets_both"] and (best_plan is None or plan["cost"] < best_plan["cost"]):
            best_plan = plan
    if best_plan is None: best_plan = plan
    best_plan["time"] = t_total
    best_plan["dual_history"] = history
    return best_plan

print("=== SMALL ===")
t0 = time.time()
h_s = dual_averaging_loop(inst_s, max_outer=3)
t_s = time.time() - t0
print(f"Hybrid SMALL: cost=${h_s['cost']:.0f} SLA={h_s['sla']:.1%} ASA={h_s['asa']:.1f}s "
      f"both={h_s['meets_both']} tag={h_s.get('tag')} t={t_s:.2f}s")
print(f"Greedy SMALL: cost=${g_s['cost']:.0f} SLA={g_s['sla']:.1%} ASA={g_s['asa']:.1f}s both={g_s['meets_both']}")
print(f"Δ=${h_s['cost']-g_s['cost']:.0f}  beat={h_s['meets_both'] and g_s['meets_both'] and h_s['cost']<g_s['cost']-0.5}")

print("\n=== LARGE ===")
t0 = time.time()
h_l, _ = run_cvar_qaoa(inst_l, alpha=0.25, shots=400, maxiter=3)
t_l = time.time() - t0
print(f"Hybrid LARGE: cost=${h_l['cost']:.0f} SLA={h_l['sla']:.1%} ASA={h_l['asa']:.1f}s "
      f"both={h_l['meets_both']} tag={h_l.get('tag')} t={t_l:.2f}s")
print(f"Greedy LARGE: cost=${g_l['cost']:.0f} SLA={g_l['sla']:.1%} ASA={g_l['asa']:.1f}s both={g_l['meets_both']}")
print(f"Δ=${h_l['cost']-g_l['cost']:.0f}  beat={h_l['meets_both'] and g_l['meets_both'] and h_l['cost']<g_l['cost']-0.5}")
```

## 9. Manager controls


### Sensitivity analysis theory — re-solving under changed preferences/constraints

The four manager scenarios are small, principled perturbations of the same QUBO:

* **`cost`** — the baseline objective, no changes.
* **`preference`** — reweights $c_s\to c_s\cdot m_s$ with shift multipliers $m=(0.9,1.0,1.15,\ldots)$,
  i.e. a *linear* re-weighting of the QUBO's linear term only (the $Q_{ij}$ quadratic block is
  untouched) — a cheap way to explore "prefer morning shifts" type policies without re-deriving
  the whole model.
* **`resilience`** — tightens the RHS $R_t^{\mathrm{SLA}}, R_t^{\mathrm{ASA}}\to R_t+1$ (capped
  at 40), i.e. a **constraint-tightening** sensitivity analysis: how much does one extra unit of
  safety margin per interval cost?

Each scenario is solved independently by a fresh (non-warm-started-by-`resilience`,
non-outer-loop) `run_cvar_qaoa` call — a single-shot QAOA solve per scenario, not the full
Lagrangian outer loop, trading a small amount of dual-feasibility-seeking for four independent
answers quickly.

```python

def manager_scenario(inst, mode="cost"):
    inst2 = dict(inst)
    costs = inst["costs"].copy()
    if mode == "preference":
        costs = costs * np.array([0.9, 1.0, 1.15][:inst["n_shifts"]] + [1.0]*max(0, inst["n_shifts"]-3))
        inst2["costs"] = costs
    if mode == "resilience":
        inst2["R_sla"] = np.minimum(inst["R_sla"] + 1, 40)
        inst2["R_asa"] = np.minimum(inst["R_asa"] + 1, 40)
    plan, _ = run_cvar_qaoa(inst2, alpha=0.25, shots=280, maxiter=2)
    plan["mode"] = mode
    return plan

print("Manager scenarios (SMALL):")
sc_rows = []
for mode in ["cost", "service", "preference", "resilience"]:
    sc = manager_scenario(inst_s, mode)
    sc_rows.append(dict(mode=mode, cost=sc["cost"], sla=sc["sla"], asa=sc["asa"], both=sc["meets_both"]))
    print(f"  {mode:12s}  cost=${sc['cost']:.0f}  SLA={sc['sla']:.1%}  ASA={sc['asa']:.1f}s  both={sc['meets_both']}")
sdf = pd.DataFrame(sc_rows)
display(sdf)

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(sdf["mode"], sdf["cost"], color="#DD8452")
ax.set_ylabel("Cost ($)"); ax.set_title("Manager control scenarios — cost")
plt.tight_layout(); plt.show()
```

## 10. Simulation detail (abandonment, utilization)


### Little's Law context for abandonment & utilization

The abandonment/occupancy metrics reported here are consistent with the queueing-theoretic
picture underlying Erlang-C: by **Little's Law**, the average number of calls in the system
$L=\lambda W$ (arrival rate times average time in system). High occupancy
($\mathrm{Occ}\to1$, i.e. $A\to n$) drives both $W$ (wait time, hence ASA) and abandonment
upward nonlinearly (visible in the log-scaled ASA/abandonment curves in §2) — so the
"Abandonment vs Utilization" pair of bars below is really reporting two views of the same
underlying occupancy regime each method's staffing plan lands in.
